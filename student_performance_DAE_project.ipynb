```python
# <---------- CHECKING INCORRECT ATTENDANCE VALUES ---------->

invalid_attendance = df[
    (df['Attendance (%)'] < 0) |
    (df['Attendance (%)'] > 100)
]

print("Incorrect Attendance Records:")
print(invalid_attendance)
```

```python
# <---------- CHECKING INCORRECT MIDTERM SCORES ---------->

invalid_midterm = df[
    (df['Midterm_Score'] < 0) |
    (df['Midterm_Score'] > 100)
]

print("Incorrect Midterm Score Records:")
print(invalid_midterm)
```

```python
# <---------- CHECKING INCORRECT FINAL SCORES ---------->

invalid_final = df[
    (df['Final_Score'] < 0) |
    (df['Final_Score'] > 100)
]

print("Incorrect Final Score Records:")
print(invalid_final)
```

```python
# <---------- REMOVING DUPLICATE RECORDS ---------->

df = df.drop_duplicates()

print("Duplicate records removed successfully.")
```

```python
# <---------- VERIFYING DATA AFTER CLEANING ---------->

print("Remaining Duplicate Records:", df.duplicated().sum())
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
#<---------- MODULE 4: DATA PREPROCESSING ---------->
# <---------- PREPARING NUMERICAL DATA ---------->

numeric_columns = df.select_dtypes(
    include=np.number
).columns

print("Numerical Columns:")
print(numeric_columns.tolist())
```

```python
# <---------- CHECKING NUMERICAL DATA ---------->

print("Numerical Data Summary:")
print(df[numeric_columns].describe())
```

```python
# <---------- DETECTING OUTLIERS USING IQR ---------->

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(column, "-> Outliers:", len(outliers))
```

```python
# <---------- DETECTING OUTLIERS USING IQR ---------->

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(column, "-> Outliers:", len(outliers))
```

```python
# <---------- CREATING AVERAGE FOR PERFORMANCE ANALYSIS ---------->

df['Average'] = df[
    ['Midterm_Score', 'Final_Score']
].mean(axis=1)

print("Average column created successfully.")

df[
    ['Student_ID', 'Midterm_Score', 'Final_Score', 'Average']
].head()
```

```python
# <---------- VERIFYING PREPROCESSED DATA ---------->

print("Dataset Shape:", df.shape)
print("Total Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Records:", df.duplicated().sum())
```

```python
#<---------- MODULE 5: PERFORMANCE ANALYSIS ---------->
# <---------- CALCULATING STUDENT AVERAGE ---------->

df['Average'] = df[
    ['Midterm_Score', 'Final_Score']
].mean(axis=1)

print("Average calculated successfully!")

df[
    ['Student_ID', 'First_Name', 'Last_Name',
     'Midterm_Score', 'Final_Score', 'Average']
].head()
```

```python
# <---------- CHECKING EXISTING GRADES ---------->

print("Grade Categories:")
print(df['Grade'].unique())

print("\nGrade Distribution:")
print(df['Grade'].value_counts())
```

```python
# <---------- GENERATING CALCULATED GRADE ---------->

def calculate_grade(average):
    if average >= 90:
        return 'A+'
    elif average >= 80:
        return 'A'
    elif average >= 70:
        return 'B'
    elif average >= 60:
        return 'C'
    elif average >= 50:
        return 'D'
    else:
        return 'F'

df['Calculated_Grade'] = df['Average'].apply(calculate_grade)

print("Calculated Grade generated successfully!")

df[
    ['Student_ID', 'Average', 'Grade', 'Calculated_Grade']
].head()
```

```python
# <---------- SORTING STUDENTS BY PERFORMANCE ---------->

df_sorted = df.sort_values(
    by='Average',
    ascending=False
)

print("Students Sorted by Average Score:")

df_sorted[
    ['Student_ID', 'First_Name', 'Last_Name',
     'Average', 'Grade']
].head(10)
```

```python
# <---------- FILTERING HIGH PERFORMING STUDENTS ---------->

high_performers = df[
    df['Average'] >= 75
]

print("High Performing Students:")

high_performers[
    ['Student_ID', 'First_Name', 'Last_Name', 'Average', 'Grade']
]
```

```python
# <---------- IDENTIFYING TOPPER ---------->

topper_index = df['Average'].idxmax()

topper = df.loc[topper_index]

print("Topper Information:")
print("Student ID:", topper['Student_ID'])
print("Name:", topper['First_Name'], topper['Last_Name'])
print("Average:", topper['Average'])
print("Grade:", topper['Grade'])
```

```python
# <---------- DEPARTMENT-WISE PERFORMANCE ANALYSIS ---------->

department_performance = df.groupby(
    'Department'
)['Average'].mean().sort_values(
    ascending=False
)

print("Department-wise Average Performance:")
print(department_performance)
```

```python
# <---------- MIDTERM AND FINAL PERFORMANCE COMPARISON ---------->

print("Average Midterm Score:",
      df['Midterm_Score'].mean())

print("Average Final Score:",
      df['Final_Score'].mean())

if df['Final_Score'].mean() > df['Midterm_Score'].mean():
    print("Overall final performance is higher than midterm performance.")
elif df['Final_Score'].mean() < df['Midterm_Score'].mean():
    print("Overall midterm performance is higher than final performance.")
else:
    print("Midterm and final performance are equal.")
```

```python
# <---------- FINAL PERFORMANCE ANALYSIS ---------->

print("Total Students:", len(df))
print("Average Student Score:", round(df['Average'].mean(), 2))
print("Highest Average Score:", round(df['Average'].max(), 2))
print("Lowest Average Score:", round(df['Average'].min(), 2))

print("\nPerformance Analysis Completed Successfully!")
```

```python
#<---------- MODULE 6: VISUALIZATION ---------->
# <---------- GRADE DISTRIBUTION BAR GRAPH ---------->

grade_counts = df['Grade'].value_counts()

plt.figure(figsize=(8, 5))

grade_counts.plot(kind='bar')

plt.title("Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)

plt.show()
```

```python
# <---------- GRADE DISTRIBUTION PIE CHART ---------->

grade_counts = df['Grade'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct='%1.1f%%'
)

plt.title("Grade Distribution of Students")

plt.show()
```

```python
# <---------- MIDTERM VS FINAL PERFORMANCE COMPARISON ---------->

average_scores = [
    df['Midterm_Score'].mean(),
    df['Final_Score'].mean()
]

subjects = ['Midterm Score', 'Final Score']

plt.figure(figsize=(8, 5))

plt.bar(subjects, average_scores)

plt.title("Midterm vs Final Performance")
plt.xlabel("Assessment")
plt.ylabel("Average Score")

plt.show()
```

```python
# <---------- DEPARTMENT-WISE PERFORMANCE COMPARISON ---------->

department_average = df.groupby(
    'Department'
)['Average'].mean()

plt.figure(figsize=(10, 5))

department_average.plot(kind='bar')

plt.title("Department-wise Average Performance")
plt.xlabel("Department")
plt.ylabel("Average Score")
plt.xticks(rotation=45)

plt.show()
```

```python
# <---------- STUDENT PERFORMANCE COMPARISON ---------->

top_students = df.sort_values(
    by='Average',
    ascending=False
).head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    top_students['Student_ID'].astype(str),
    top_students['Average']
)

plt.title("Top 10 Student Performance")
plt.xlabel("Student ID")
plt.ylabel("Average Score")
plt.xticks(rotation=45)

plt.show()
```
